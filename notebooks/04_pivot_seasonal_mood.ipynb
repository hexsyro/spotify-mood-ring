## 📓 Notebook 04: Pivot & Reshape

**Skills practiced:** `pivot_table(aggfunc='mean')` vs `groupby().mean()`, 
`melt()` round-trips, grouping by decade.

**Goal:** Build two pivot tables — mean valence by year × genre/rank-bucket, 
and mean energy by decade × rank-bucket — then confirm a `melt()` round-trip 
matches the expected row count.

```python
import pandas as pd
import numpy as np

df=pd.read_csv('../data/processed/merged_data.csv')

#pivot the data to get the mean valence by year and track genre
pivoted = df.pivot_table(index='year', columns='track_genre', values='valence', aggfunc='mean').round(3)

#melt the pivoted data to get a long format dataframe with year, track_genre, and valence columns
pivoted = pivoted.reset_index()
melted = pivoted.melt(id_vars='year', var_name='track_genre', value_name='valence').dropna()
melted.to_csv('../data/processed/valence_by_genre_data.csv', index=False)
print(melted.head())
```

```python

#pivoting the data to get the mean energy by decade and rank
decade=(df['year']//10*10).rename('decade')
pivoted2 = df.pivot_table(index=decade, columns='rank', values='energy', aggfunc='mean').round(3)

#melting the data to get the mean energy by decade and rank in a long format
pivoted2=pivoted2.reset_index()
melted2=pivoted2.melt(id_vars=['decade'], value_vars=[x for x in pivoted2.columns if x != 'decade'], value_name='energy', var_name='rank').dropna()
melted2.to_csv('../data/processed/energy_by_rank_data.csv', index=False)

print(melted2.head())
```
