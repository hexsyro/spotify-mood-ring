## 📓 Notebook 03: Filtering & Questions

**Skills practiced:** Boolean indexing, combining conditions with `&`, 
`value_counts()`.

**Goal:** Answer four concrete questions about the merged dataset using 
single-line filters — top-10 songs, lowest-valence songs, most-recurring 
artists, and high-energy songs from a specific year.

```python
import pandas as pd

df=pd.read_csv("../data/processed/merged_data.csv")
```

```python
print(f"All songs that ranked in the top 10 in any year :\n {df[df['rank'] <= 10]['title'].drop_duplicates().tolist()}.")
```

```python
#.loc[lambda x: x > 3]: This is a clean Pandas trick to filter our grouped data on the fly, keeping only the records where the unique year count is strictly greater than 3.

print(f" All artists that appear in more than 3 different years :\n {df.groupby('artists')['year'].nunique().loc[lambda x: x > 3].index.tolist()}.")
```

```python
print(f"All songs with a valence score below 0.3 (the 'saddest' songs) : \n {df[df['valence'] < 0.3]['title'].drop_duplicates().tolist()}")
```

```python
print(f"All songs from a single specific year AND with energy above 0.8 : \n {df[(df['year'] == 2020) & (df['energy'] > 0.8)]['title'].drop_duplicates().tolist()}")
```
